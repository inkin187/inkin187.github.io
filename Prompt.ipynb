以下是具体要求：

角色：前端开发工程师，擅长 React + Vite，审美符合高端设计师作品集，严格克制，拒绝模板化效果

任务：从零开发 一个AI产品工程师 / AI agent工程师个人作品集网站，先交付【可直接运行预览的基础可用版本】，后续我会补充简历文本、作品图、参考网站、参考截图持续迭代优化，本次不需要填充真实简历内容，只预留好内容插槽。

视觉核心要求：
整体暗色系，风格高级、克制、极简科技感；❌ 禁止廉价霓虹光效、漫天粒子、杂乱悬浮动画、弹窗特效、花里胡哨的转场，拒绝市面上同质化作品集模板风格；科技感靠排版、留白、细微线条、低透明度渐变、精细hover微交互体现，而不是大量特效。
主版心：PC端版心最大宽度限制 1700px，内容居中；超过1700px的大屏只留白，不拉伸内容；本次优先保障PC端体验，移动端做基础降级（不强制精细适配，保证内容可浏览即可）

页面从上到下固定模块顺序：
1. 全屏Hero首页模块
- 全屏高度布局；背景支持视频，必须增加兜底：视频加载失败/关闭自动播放时自动切换深色静态渐变背景
- 包含：顶部固定导航栏（导航项：首页、精选项目、个人经历、能力优势、联系我）、主视觉大标题（设计师身份slogan预留插槽）、右上角联系CTA按钮
- 视频要求：静音、自动播放、循环、不抢占音频；给视频加低透明度深色遮罩，保证文字可读性

2. 精选项目模块
- 大卡片布局展示作品；卡片预留作品图片、项目标题、简短描述插槽；hover仅做轻微缩放/亮度微调，禁止夸张动画；卡片网格布局

3. 个人经历模块
- 布局包含：头像/人像图位、个人简介文本插槽、项目成果数据插槽、基础联系信息插槽

4. 个人优势模块
- 卡片式布局展示能力标签/能力介绍，卡片简约，少装饰

5. 全屏底部联系收尾模块（整屏高度，作为页面收尾）
- 整屏高度，简洁，放置完整联系方式、邮箱、社交链接预留位置，干净收尾
